```python
# Simple File Manager in Python

import os

while True:
    print("\n===== FILE MANAGER =====")
    print("1. Create File")
    print("2. Read File")
    print("3. Delete File")
    print("4. Rename File")
    print("5. List Files")
    print("6. Exit")

    choice = input("Enter your choice: ")

    # Create File
    if choice == "1":
        filename = input("Enter file name: ")

        with open(filename, "w") as file:
            content = input("Enter content for file: ")
            file.write(content)

        print("File created successfully!")

    # Read File
    elif choice == "2":
        filename = input("Enter file name to read: ")

        if os.path.exists(filename):
            with open(filename, "r") as file:
                print("\nFile Content:\n")
                print(file.read())
        else:
            print("File does not exist!")

    # Delete File
    elif choice == "3":
        filename = input("Enter file name to delete: ")

        if os.path.exists(filename):
            os.remove(filename)
            print("File deleted successfully!")
        else:
            print("File does not exist!")

    # Rename File
    elif choice == "4":
        old_name = input("Enter current file name: ")
        new_name = input("Enter new file name: ")

        if os.path.exists(old_name):
            os.rename(old_name, new_name)
            print("File renamed successfully!")
        else:
            print("File does not exist!")

    # List Files
    elif choice == "5":
        files = os.listdir()

        print("\nFiles in Current Folder:\n")

        for file in files:
            print(file)

    # Exit
    elif choice == "6":
        print("Exiting File Manager...")
        break

    else:
        print("Invalid choice! Try again.")
```

```python
# rent calculator 
# Rent Calculator in Python

print("===== RENT CALCULATOR =====")

# Inputs
room_rent = int(input("Enter room rent: "))
food_bill = int(input("Enter food bill: "))
electricity_units = int(input("Enter electricity units used: "))
charge_per_unit = int(input("Enter charge per unit: "))
persons = int(input("Enter number of persons: "))

# Electricity bill
electricity_bill = electricity_units * charge_per_unit

# Total expense
total_bill = room_rent + food_bill + electricity_bill

# Per person amount
per_person = total_bill / persons

# Output
print("\n===== BILL DETAILS =====")
print(f"Room Rent: ₹{room_rent}")
print(f"Food Bill: ₹{food_bill}")
print(f"Electricity Bill: ₹{electricity_bill}")
print(f"Total Bill: ₹{total_bill}")
print(f"Each Person Should Pay: ₹{per_person}")
```

```python
# Student Management System in Python

students = {}

while True:
    print("\n===== STUDENT MANAGEMENT SYSTEM =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update Student")
    print("5. Delete Student")
    print("6. Count Students")
    print("7. Exit")

    choice = input("Enter your choice: ")

    # Add Student
    if choice == "1":
        roll_no = input("Enter Roll Number: ")

        if roll_no in students:
            print("Student already exists!")
        else:
            name = input("Enter Name: ")
            age = input("Enter Age: ")
            course = input("Enter Course: ")

            students[roll_no] = {
                "Name": name,
                "Age": age,
                "Course": course
            }

            print("Student added successfully!")

    # View Students
    elif choice == "2":
        if not students:
            print("No student records found!")
        else:
            print("\n===== STUDENT RECORDS =====")

            for roll_no, details in students.items():
                print(f"\nRoll No: {roll_no}")
                print(f"Name: {details['Name']}")
                print(f"Age: {details['Age']}")
                print(f"Course: {details['Course']}")

    # Search Student
    elif choice == "3":
        roll_no = input("Enter Roll Number to search: ")

        if roll_no in students:
            print("\nStudent Found!")
            print(f"Name: {students[roll_no]['Name']}")
            print(f"Age: {students[roll_no]['Age']}")
            print(f"Course: {students[roll_no]['Course']}")
        else:
            print("Student not found!")

    # Update Student
    elif choice == "4":
        roll_no = input("Enter Roll Number to update: ")

        if roll_no in students:
            name = input("Enter New Name: ")
            age = input("Enter New Age: ")
            course = input("Enter New Course: ")

            students[roll_no] = {
                "Name": name,
                "Age": age,
                "Course": course
            }

            print("Student updated successfully!")
        else:
            print("Student not found!")

    # Delete Student
    elif choice == "5":
        roll_no = input("Enter Roll Number to delete: ")

        if roll_no in students:
            del students[roll_no]
            print("Student deleted successfully!")
        else:
            print("Student not found!")

    # Count Students
    elif choice == "6":
        print(f"Total Students: {len(students)}")

    # Exit
    elif choice == "7":
        print("Exiting Student Management System...")
        break

    else:
        print("Invalid choice! Please try again.")
```

```python
# TO-DO List App in Python

tasks = []

while True:
    print("\n===== TO-DO LIST APP =====")
    print("1. Add Task")
    print("2. View Tasks")
    print("3. Remove Task")
    print("4. Clear All Tasks")
    print("5. Exit")

    choice = input("Enter your choice: ")

    # Add Task
    if choice == "1":
        task = input("Enter new task: ")
        tasks.append(task)
        print("Task added successfully!")

    # View Tasks
    elif choice == "2":
        if not tasks:
            print("No tasks available!")
        else:
            print("\n===== YOUR TASKS =====")

            for index, task in enumerate(tasks, start=1):
                print(f"{index}. {task}")

    # Remove Task
    elif choice == "3":
        if not tasks:
            print("No tasks to remove!")
        else:
            print("\n===== TASKS =====")

            for index, task in enumerate(tasks, start=1):
                print(f"{index}. {task}")

            task_no = int(input("Enter task number to remove: "))

            if 1 <= task_no <= len(tasks):
                removed_task = tasks.pop(task_no - 1)
                print(f"Task '{removed_task}' removed successfully!")
            else:
                print("Invalid task number!")

    # Clear All Tasks
    elif choice == "4":
        tasks.clear()
        print("All tasks cleared!")

    # Exit
    elif choice == "5":
        print("Exiting TO-DO List App...")
        break

    else:
        print("Invalid choice! Please try again.")
```
